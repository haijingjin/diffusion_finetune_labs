```python
from diffusers import StableDiffusionPipeline
import torch
from pathlib import Path
from PIL import Image

# Load base model
pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16
).to("cuda")

# Load LoRA weights
pipe.unet.load_attn_procs("/content/drive/MyDrive/hg_imGen/LoRA_fine_tuned_model")

# Prompt
prompt = "a fox under a sakura tree, in (hj_goph style)"

# Output folder
output_dir = Path("/content/drive/MyDrive/hg_imGen/LoRA_gen_output")
output_dir.mkdir(parents=True, exist_ok=True)

# Generate multiple images
for i in range(10):
    image = pipe(prompt, num_inference_steps=50, guidance_scale=7.5).images[0]
    image.save(output_dir / f"lora_fox_{i+1}.png")

print("Done! Images saved.")
```
