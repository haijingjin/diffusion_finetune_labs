```python
from diffusers import StableDiffusionPipeline
import torch
from pathlib import Path
import datetime

# Load fine-tuned model
pipe = StableDiffusionPipeline.from_pretrained(
    "/content/drive/MyDrive/hg_imGen/pretraining_output",
    torch_dtype=torch.float16
).to("cuda")

# Define prompt and number of images
prompt = "a photo of [sks_hj] fox under a sakura tree, pastel colors"
num_images = 10

# Define output directory
output_dir = Path("/content/drive/MyDrive/hg_imGen/pretrained_fox_images")
output_dir.mkdir(parents=True, exist_ok=True)

# Generate and save
for i in range(num_images):
    image = pipe(prompt).images[0]
    timestamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
    filename = output_dir / f"pretraind_sks_hj_fox_sakura_{timestamp}_{i+1}.png"
    image.save(filename)
    print(f"Saved: {filename}")
```
